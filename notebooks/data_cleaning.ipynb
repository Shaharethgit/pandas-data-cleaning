# 🟠 LEVEL 8 — Outlier Detection

# Task 26 — Detect Salary Outliers

**Objective:   Identify unusually low or high salaries using the Interquartile Range (IQR) method.**

**Approach:**
1. Calculate the first quartile (`Q1`) using the 25th percentile.
2. Calculate the third quartile (`Q3`) using the 75th percentile.
3. Calculate the IQR.
4. Calculate the lower and upper outlier boundaries.
5. Create a new column named `salary_outlier` to identify potential outliers.

**Formulas:**
- `IQR = Q3 - Q1`
- `Lower Boundary = Q1 - (1.5 × IQR)`
- `Upper Boundary = Q3 + (1.5 × IQR)`

**Expected result:**
- `True` — salary is below the lower boundary or above the upper boundary.
- `False` — salary is within the boundaries.

**Important:   Identify potential outliers without immediately deleting or modifying the original salary values.**

````python
Q1 = df["salary"].quantile(0.25)
Q3 = df["salary"].quantile(0.75)
IQR = Q3-Q1
lower = Q1 - (1.5 * IQR)
upper = Q3 + (1.5 * IQR)
df["salary_outlier"] = ((df["salary"] < lower) | (df["salary"] > upper)).where(df["salary"].notna(), np.nan)
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower boundary:", lower)
print("Upper boundary:", upper)
df.head(10)
````
