# 🟠 LEVEL 8 — Outlier Detection

# Task 26 — Detect Salary Outliers

**Objective:   Identify unusually low or high salaries using the Interquartile Range (IQR) method.**

**Approach:**
1. **Calculate the first quartile (`Q1`) using the 25th percentile.**
2. **Calculate the third quartile (`Q3`) using the 75th percentile.**
3. **Calculate the IQR.**
4. **Calculate the lower and upper outlier boundaries.**
5. **Create a new column named `salary_outlier` to identify potential outliers.**

**Formulas:**
- **`IQR = Q3 - Q1`**
- **`Lower Boundary = Q1 - (1.5 × IQR)`**
- **`Upper Boundary = Q3 + (1.5 × IQR)`**

**Expected result:**
- **`True` — salary is below the lower boundary or above the upper boundary.**
- **`False` — salary is within the boundaries.**

**Important:   Identify potential outliers without immediately deleting or modifying the original salary values.**

````python
Q1 = df["salary"].quantile(0.25)
Q3 = df["salary"].quantile(0.75)
IQR = Q3-Q1
lower = Q1 - (1.5 * IQR)
upper = Q3 + (1.5 * IQR)
df["salary_outlier"] = ((df["salary"] < lower) | (df["salary"] > upper)).where(df["salary"].notna(), np.nan)
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower boundary:", lower)
print("Upper boundary:", upper)
df.head(10)
````

# Task 35 — Create Salary Band

**Objective:   Categorize employees into salary bands based on their salary.**

**Rules:**
- **Salary below `6000` → `Low`**
- **Salary from `6000` to `7999` → `Medium`**
- **Salary `8000` or above → `High`**

**Tasks:**
- **Define conditions for each salary band.**
- **Use `np.select()` to assign the appropriate label.**
- **Store the result in a new column named `salary_band`.**

**Expected Result:**
- **Each valid salary is assigned to `Low`, `Medium`, or `High`.**
- **Missing salaries remain missing.**

````python
#Method 1 — np.select()
category = (
    df["salary"] < 6000,
    df["salary"].between(6000,7999, inclusive="both"),
    df["salary"] >= 8000
)
labels = ['Low', 'Medium', 'High']
df["salary_band"] = np.select(category, labels, default=pd.NA)

#Method 2 — pd.cut()
df["salary_band2"] = pd.cut(
    df["salary"],
    bins=[-float("inf"), 6000, 8000, float("inf")],
    labels=["Low", "Medium", "High"],
    right=False
)
df.head(10)
````

# Task 38 — Create a Complete Data Quality Report

**Objective:  Create a summary table showing missing, invalid, and duplicate values for selected employee data columns.**

**Report Metrics:**
- **`missing`: Number of missing values in each column.**
- **`invalid`: Number of values failing the available validation checks.**
- **`duplicates`: Number of repeated values within each column.**

**Tasks:**
- **Count missing values using `isna().sum()`.**
- **Count invalid values using the existing validation flags for age, salary, bonus, email, phone, and join date.**
- **Count repeated values within each column using `duplicated().sum()`.**
- **Combine the results into a single summary DataFrame.**

**Important Note:  Invalid counts for `name`, `department`, `gender`, and `status` remain zero because separate validation checks have not been defined for these columns.**

**Expected Result:**
- **All 11 selected columns appear in the report.**
- **Missing, invalid, and duplicate counts are displayed separately.**
- **Invalid counts reflect the available validation checks.**

````python
summary_columns = ['employee_id','name','department','age','salary','bonus','join_date','email','phone','gender','status']

missing = df[summary_columns].isna().sum()

invalid = pd.Series(0, index = summary_columns, dtype = "int")
columns = ['age_valid','salary_valid','bonus_valid','email_valid','phone_valid','join_date_valid']

for cl in columns:
     count = (~df[cl]).sum()
     if cl == 'age_valid':
          invalid['age'] = count
     elif cl == 'salary_valid' :
          invalid['salary'] = count
     elif cl == 'bonus_valid' :
               invalid['bonus'] = count
     elif cl == 'email_valid' :
               invalid['email'] = count
     elif cl == 'phone_valid' :
               invalid['phone'] = count
     else :
               invalid['join_date'] = count

duplicates = pd.Series(0, index = summary_columns, dtype = "int")
for cl in summary_columns :
        count = df[cl].duplicated().sum()
        duplicates[cl] = count

summary = pd.DataFrame({
        "missing" : missing,
        "invalid" : invalid,
        "duplicates" : duplicates
})
summary.reset_index(names = "Column_name")
````

# 🟢 Bonus Task — Handle Missing Values

**Objective:   Handle missing values using appropriate strategies based on the meaning and type of each column.**

**Approach:**
- **Bonus:   Fill missing values with `0`, assuming a missing bonus means no bonus was paid.**
- **Age:   Fill missing values with the median age.**
- **Salary:   Fill missing values with the median salary.**
- **Department, Gender, Status:   Fill missing values with the mode (most frequent value), when available.**

**Important Note:   Missing-value handling depends on business meaning. Zero is appropriate for bonus only when a missing value genuinely means no bonus. Median and mode imputation are choices that should be justified for the dataset.**

**Expected Result:   Missing values in the selected columns are handled according to the chosen rules.**

````python
df["bonus"] = df["bonus"].fillna(0)
df["age"] = df["age"].fillna(df["age"].median())
df["salary"] = df["salary"].fillna(df["salary"].median())

for cl in ["department", "gender", "status"]:
    if df[cl].notna().any():
        df[cl] = df[cl].fillna(df[cl].mode()[0])
````
