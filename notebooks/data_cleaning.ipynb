# Task 2 — Clean column names
### Convert: " Employee ID ", "NAME", "Department", "Join Date" Into: employee_id, name, department, join_date

```python
df.columns = df.columns.str.strip()
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(r"\s+","_", regex=True)
df.columns
```

# Task 3 — Remove leading/trailing whitespace
### Clean whitespace from all string columns.
### For example:, "     John Smith             ", "it                 ", "    2022-05-20             ", should become: "John Smith", "it", "2022-05-20"

```python
for cl in df.select_dtypes(include="str").columns :
    df[cl] = df[cl].str.strip()
    df[cl] = df[cl].str.replace(r"\s+", " ", regex = True)    
df.head(10)
```

# Task 4 — Find completely empty strings
### Find cells containing:   ""   ,   "   "   and convert them to: NaN

```python
for cl in df.columns:
    df[cl] = df[cl].replace(r"^\s*$", np.nan, regex=True)
df.head(10)
```

# 🟢 LEVEL 2 — Missing Values
# Task 5 — Identify missing values
### Find missing values: df.isna() Then calculate missing count for every column

```python
missing_values = df.isna().sum()
missing_values
```

# Task 6 — Missing percentage
### Calculate: columnmissing_count , missing_percentage 

```python
missing_percentage = df.isna().mean() * 100.0
result = pd.DataFrame({
    "missing_count" : missing_values,
    "missing_percentage" : missing_percentage
})
result = result.reset_index(names="Column")
result
```

# Task 7 — Replace fake missing values
### Convert these into actual NaN:  "N/A", "NULL" ,"unknown" , "Unknown" , ""

```python
fake_missing = ["N/A","NULL","unknown","Unknown","", "null"]
for cl in df.columns:
    df[cl] = df[cl].replace(fake_missing,np.nan)
df.head(10)
```
