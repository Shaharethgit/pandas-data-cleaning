# 🟢 LEVEL 1 — Basic Structural Cleaning

## Task 2 — Clean Column Names

**Convert column names into lowercase, remove extra spaces, and replace spaces between words with underscores.**

**Example:**
- `" Employee ID "` → `employee_id`
- `"NAME"` → `name`
- `"Department"` → `department`
- `"Join Date"` → `join_date`

````python
df.columns = df.columns.str.strip()
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(r"\s+","_", regex=True)
df.columns
````

# 🟢 LEVEL 1 — Basic Structural Cleaning

## Task 3 — Remove Leading and Trailing Whitespace

**Clean leading and trailing whitespace from all string columns.**

**Examples:**
- `"     John Smith     "` → `"John Smith"`
- `"it          "` → `"it"`
- `"    2022-05-20     "` → `"2022-05-20"`

````python
for cl in df.select_dtypes(include="str").columns :
    df[cl] = df[cl].str.strip()
    df[cl] = df[cl].str.replace(r"\s+", " ", regex = True)    
df.head(10)
````

# 🟢 LEVEL 1 — Basic Structural Cleaning

## Task 4 — Find Completely Empty Strings

- **Find cells containing empty strings (`""`) or strings with only whitespace (`"   "`).**
- **Replace them with `NaN`.**

````python
for cl in df.columns:
    df[cl] = df[cl].replace(r"^\s*$", np.nan, regex=True)
df.head(10)
````

# 🟢 LEVEL 2 — Missing Values

## Task 5 — Identify Missing Values

- **Identify missing values using `df.isna()`.**
- **Calculate the missing value count for every column.**

````python
missing_values = df.isna().sum()
missing_values
````

# 🟢 LEVEL 2 — Missing Values

## Task 6 — Calculate Missing Percentage

- **Calculate the missing value count for every column.**
- **Calculate the missing percentage for every column.**

````python
missing_percentage = df.isna().mean() * 100.0
result = pd.DataFrame({
    "missing_count" : missing_values,
    "missing_percentage" : missing_percentage
})
result = result.reset_index(names="Column")
result
````

# 🟢 LEVEL 2 — Missing Values

## Task 7 — Replace Fake Missing Values

**Convert the following fake missing values into actual `NaN`:**

- **`"N/A"`**
- **`"NULL"`**
- **`"unknown"`**
- **`"Unknown"`**
- **`""`** (empty string)

````python
fake_missing = ["N/A","NULL","unknown","Unknown","", "null"]
for cl in df.columns:
    df[cl] = df[cl].replace(fake_missing,np.nan)
df.head(10)
````

## 🟢 LEVEL 2 — Missing Values

# Task 8 — Decide How to Handle Missing Values

**Missing-value treatment will be performed after completing the data-cleaning and data-type conversion tasks to avoid affecting subsequent analysis.**

**The appropriate strategy will be selected based on each column's meaning and the amount of missing data.**

## Task 9 — Create Missing Indicators

**Create the following columns:**
- **`age_missing`**
- **`salary_missing`**
- **`email_missing`**
- **`phone_missing`**

**Example:**
- **`age = NaN` → `age_missing = 1`**
- **`age = 25` → `age_missing = 0`**

````python
missing_columns = ['age','salary','email','phone']
for cl in missing_columns:
    df[cl+"_missing"] = df[cl].isna().astype("int")
df.head(10)
````

# 🟡 LEVEL 3 — Data Types

## Task 10 — Convert Age
**Currently:**
- **`"25"`**
- **`" 30 "`**
- **`"35"`**
- **`"twenty-eight"`**
- **`"N/A"`**
**Objective: Convert `age` into numeric values.**
**Expected result: Invalid values should become `NaN`**


````python
df["age"] = pd.to_numeric(df["age"], errors="coerce")
df.head(10)
````

# 🟡 LEVEL 3 — Data Types

## Task 11 — Convert Salary

**Currently:**
- **`"$5,000"`**
- **`"6000"`**
- **`"7,500.50"`**
- **`"8,000"`**

**Objective: Remove currency symbols and commas, then convert `salary` into numeric values.**

**Expected result:**
- **`5000`**
- **`6000`**
- **`7500.50`**
- **`8000`**

````python
df["salary"] = df["salary"].str.replace('$',"", regex=False)
df["salary"] = df["salary"].str.replace(',',"", regex=False)
df["salary"] = pd.to_numeric(df["salary"], errors="coerce")
df.head(10)
````

# 🟡 LEVEL 3 — Data Types

## Task 12 — Convert Bonus to Numeric

**Clean the `bonus` column and convert it to a numeric data type.**

**Example values:**
- **`"$750"`**
- **`"1000"`**
- **`"-200"`**
- **`"unknown"`**

**Expected result:**
- **`"$750"` → `750`**
- **`"1000"` → `1000`**
- **`"-200"` → `-200`**
- **`"unknown"` → `NaN`**

````python
df["bonus"] = df["bonus"].str.replace("$" ,"", regex=False)
df["bonus"] = df["bonus"].str.replace("," ,"" ,regex=False)
df["bonus"] = pd.to_numeric(df["bonus"], errors="coerce")
df.head(10)
````

# 🟡 LEVEL 3 — Data Types

## Task 13 — Convert Join Date

**Convert all valid dates into Pandas datetime.**

**Invalid values such as:**
```
invalid
2020/13/01
N/A
```

**should become:**
```

```

````python
df["join_date"] = pd.to_datetime(df["join_date"], errors="coerce", format="mixed", dayfirst=True)
df.head(10)
````

# 🟡 LEVEL 4 — Text Cleaning

## Task 14 — Standardize Names

**Convert names into proper case and remove multiple spaces between words.**

**Examples:**

| **Original Value** | **Cleaned Value** |
|:---|:---|
| `"JOHN SMITH"` | `"John Smith"` |
| `"john smith"` | `"John Smith"` |
| `"John Smith"` | `"John Smith"` |

# 🟡 LEVEL 4 — Text Cleaning

# Task 15 — Clean Department

**Standardize department names by fixing capitalization, extra spaces, and inconsistent department names.**

**Original Values and Expected Output:**

| **Original Value** | **Cleaned Value** |
|:---|:---|
| `IT` | `IT` |
| `it` | `IT` |
| `IT Department` | `IT` |
| `HR` | `HR` |
| `Human Resources` | `HR` |
| `Human Resource` | `HR` |
| `sales` | `Sales` |
| `Sales` | `Sales` |
| `SALES` | `Sales` |
| `Finance` | `Finance` |
| `FINANCE` | `Finance` |
| `finance` | `Finance` |
| `Operations` | `Operations` |
| `operations` | `Operations` |
| `Marketing` | `Marketing` |
| `marketing` | `Marketing` |

````python
df["department"] = df["department"].str.lower()
department_clean = {
    "it" : "IT",
    "it department" : "IT",
    "hr" : "HR",
    "human resources" : "HR",
    "human resource" : "HR",
    "sales" : "Sales",
    "finance" : "Finance",
    "operations" : "Operations",
    "marketing" : "Marketing"
}

df["department"] = df["department"].map(department_clean)

df.to_excel(r"C:\Users\User\Desktop\sqldata\r.xlsx")
df.head(10)
````
